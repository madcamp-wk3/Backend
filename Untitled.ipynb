```python
from fastapi import FastAPI
from database import db
from models import ChatMessage

app = FastAPI()

# MongoDB 컬렉션 선택
chat_collection = db["chats"]

@app.get("/")
def home():
    return {"message": "FastAPI + MongoDB Backend"}

# 메시지 저장
@app.post("/chat")
def save_message(chat: ChatMessage):
    new_chat = {
        "username": chat.username,
        "message": chat.message,
        "timestamp": chat.timestamp
    }
    chat_collection.insert_one(new_chat)
    return {"status": "success", "message": "Chat saved"}

# 메시지 불러오기
@app.get("/chat")
def get_messages():
    chats = chat_collection.find().sort("timestamp", -1)
    return [{"id": str(chat["_id"]), "username": chat["username"], "message": chat["message"], "timestamp": chat["timestamp"]} for chat in chats]

# 서버 실행 명령어: uvicorn main:app --reload
```

```python
from pymongo import MongoClient
import os
from dotenv import load_dotenv

load_dotenv()
MONGO_URI = os.getenv("MONGO_URI")
client = MongoClient(MONGO_URI)
db = client["chat_database"]

# 모델 정의
from pydantic import BaseModel
from datetime import datetime

class ChatMessage(BaseModel):
    username: str
    message: str
    timestamp: datetime = datetime.now()

# FastAPI 앱 정의
from fastapi import FastAPI

app = FastAPI()

@app.get("/")
def home():
    return {"message": "Hello, FastAPI!"}

@app.post("/chat")
def save_message(chat: ChatMessage):
    db["chats"].insert_one(chat.dict())
    return {"status": "success"}

@app.get("/chat")
def get_messages():
    chats = list(db["chats"].find())
    return [{"id": str(chat["_id"]), "username": chat["username"], "message": chat["message"], "timestamp": chat["timestamp"]} for chat in chats]
```

```python
import uvicorn

# FastAPI 앱 실행
uvicorn.run(app, host="127.0.0.1", port=8000)
```
